```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_csv("weather.csv")
print(df.head())
print(df.info())
print(df.describe())
print(df.isnull().sum())
df = df.dropna()
df['Date'] = pd.to_datetime(df['Date'])
print(df.head())

temp = df["Temperature"].to_numpy()
humidity = df["Humidity"].to_numpy()
precip = df["Precipitation"].to_numpy()

print("Temperature Mean:", np.mean(temp))
print("Temperature Median:", np.median(temp))
print("Temperature Std:", np.std(temp))

print("Humidity Mean:", np.mean(humidity))
print("Humidity Median:", np.median(humidity))
print("Humidity Std:", np.std(humidity))

print("Precipitation Mean:", np.mean(precip))
print("Precipitation Median:", np.median(precip))
print("Precipitation MeStdan:", np.std(precip))

df['Month'] = df['Date'].dt.month
monthly=df.groupby('Month')[['Temperature','Humidity','Precipitation']].mean()

print(monthly)


plt.plot(monthly.index, monthly['Temperature'], marker='o')
plt.title("Average Monthly Temperature")
plt.xlabel("Month")
plt.ylabel("Temperature")
plt.grid()
plt.show()

plt.bar(monthly.index, monthly['Precipitation'])
plt.title("Average Monthly Precipitation")
plt.xlabel("Month")
plt.ylabel("Precipitation")
#plt.grid()
plt.show()

plt.hist(df['Temperature'], bins = 10 )
plt.title("Temperature Distribution")
plt.xlabel("Temperature")
plt.ylabel("Frequency")
#plt.grid()
plt.show()
```
